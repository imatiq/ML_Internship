## 2. The question: decision, action, cost of a wrong call
Decision: Instead of reviewing 30,000 pages one at a time, a content strategist decides
per-archetype (e.g. "protect champions," "rewrite thin/stale pages").

Action: A team lead assigns each cluster a standard playbook — protect, improve, rewrite,
merge, prune, or monitor — then routes pages by cluster membership.

Cost of a wrong call: A misclassified declining page grouped with "healthy champions" gets
ignored and keeps losing visibility. A healthy page wrongly grouped with "at-risk" wastes
review time. Since clustering has no ground-truth label, the safeguard is manually inspecting
real pages from each cluster before trusting the names.

## 3. Quick look at the data (2-3 real numbers)

Three numbers below show distinct, overlapping-but-different behavior patterns already sitting in the data — evidence that clustering has something real to find.

```python
n_declining_visible = ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum()
print(f"Declining + still visible: {n_declining_visible:,} pages ({n_declining_visible/len(df):.1%})")

n_thin_visible = ((df["word_count"] < 1200) & (df["impressions_90d"] >= 250)).sum()
print(f"Thin but visible: {n_thin_visible:,} pages ({n_thin_visible/len(df):.1%})")

print(df["engagement_rate"].describe().round(2))
```
